# Converting image RGB to YUV color model

```python
import cv2
from google.colab.patches import cv2_imshow
# read image and display
image=cv2.imread("watch.jpg")
cv2_imshow(image)

# convert RGB image to YUV color model and display image
yuv_iamge=cv2.cvtColor(image,cv2.COLOR_BGR2YUV)
cv2_imshow(yuv_iamge)

# display each channel in YUV
y,u,v=cv2.split(yuv_iamge)

cv2_imshow(y)
cv2_imshow(u)
cv2_imshow(v)
```

# Image translation


```python
# Notes:
# translation is like sliding a photo on a table

import numpy as np

img=cv2.imread("watch.jpg")
num_row,num_col=img.shape[:2]
# use identity matrix so that image scale and orientation is same just move the image right and down

translation=np.float32([[1,0,70],[0,1,110]])
# add so that image will fit output screen
img_translation=cv2.warpAffine(img,translation,(num_col+70,num_row+110))
# if i want to move left and up and just give negative value
translation=np.float32([[1,0,-80],[0,1,-90]])
img_translation=cv2.warpAffine(img_translation,translation,(num_col+70+80,num_row+110+90))
cv2_imshow(img_translation)
```

# Image translation and rotation

```python


import cv2
import numpy as np


img=cv2.imread("watch.jpg")
num_row,num_col=img.shape[:2]
# move image right and down by half of image width and height
translation_matrix=np.float32([[1,0,num_col*0.5],[0,1,num_row*0.5]])
# creating rotation instruction, num_col num_row tells what should be orgin of roation
# we can also give num_col /2 and num_row_2 give angle of ratoation and sclaing
rotation_matrix=cv2.getRotationMatrix2D((num_col,num_row),30,1)

# apply image translation
img_translation=cv2.warpAffine(img,translation_matrix,(2*num_col,2*num_row))
# apply image rotation
img_rotation=cv2.warpAffine(img_translation,rotation_matrix,(2*num_col,2*num_row))


cv2_imshow(img_rotation)


```
